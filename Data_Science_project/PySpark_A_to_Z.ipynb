# Examples of Transformations and Actions in RDD

```python
from pyspark import SparkContext

sc = SparkContext.getOrCreate()

# Sample sentences
sentences = [
    "spark is fast",
    "pyspark with jupyter notebook",
    "spark rdd operations"
]

# Create RDD
rdd = sc.parallelize(sentences)
```

```python
# 1. flatMap: Splits each sentence into individual words and flattens the result into a single list of words
words_rdd = rdd.flatMap(lambda line: line.split(" "))
# Resulting RDD elements: ["spark", "is", "fast", "pyspark", "with", "jupyter", "notebook", "spark", "rdd", "operations"]

# 2. filter: Keeps only words that start with the letter 's'
filtered_rdd = words_rdd.filter(lambda word: word.startswith("s"))
# Resulting RDD elements: ["spark", "spark"]

# 3. map: Transforms each word into uppercase
mapped_rdd = filtered_rdd.map(lambda word: word.upper())
# Resulting RDD elements: ["SPARK", "SPARK"]
```

```python
result_list = mapped_rdd.collect()
print("collect():", result_list)
```

```python
total_count = mapped_rdd.count()
print("count():", total_count)
```

```python
combined_string = mapped_rdd.reduce(lambda x, y: x + " & " + y)
print("reduce():", combined_string)
```

# DataFrame with Schema

```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DoubleType

spark = SparkSession.builder.appName('CreateDataFrameSchema').getOrCreate()

data = [
    ("Alice", 28, "Engineering", 85000.50),
    ("Bob", 35, "Marketing", 62000.00),
    ("Charlie", 22, "HR", 48000.75)
]

schema = StructType([
    StructField('Name', StringType(), nullable = False),
    StructField('Age', IntegerType(), nullable = True),
    StructField('Department', StringType(), nullable = False),
    StructField('Salary', DoubleType(), nullable = True)
])


df = spark.createDataFrame(data, schema)
df.show()
df.printSchema()
```

# Filtering Dataset

Filter rows based on a numeric or string comparison:

```python
from pyspark.sql.functions import col

# Filter by numeric condition (salary greater than 80,000)
df.filter(col("salary") > 80000).show(5)

# Filter by string match (country is Germany)
df.filter(col("country") == "Germany").show(5)
```

When combining conditions in PySpark:

Use & for AND

Use | for OR

Use ~ for NOT

Important: Wrap each individual condition in parentheses ().

```python
# AND condition: Female employees in Germany
df.filter((col("gender") == "Female") & (col("country") == "Germany")).show(5)

# OR condition: Employees who are either Data Engineers or Solutions Architects
df.filter((col("job_title") == "Data Engineer") | (col("job_title") == "Solutions Architect")).show(5)

# NOT condition: Employees who are NOT from Germany
df.filter(~(col("country") == "Germany")).show(5)
```

Chain methods together to filter, select specific columns, and order the results:

```python
df.filter((col("age") >= 30) & (col("salary") > 75000)) \
  .select("first_name", "last_name", "job_title", "country", "salary") \
  .sort(col("salary").desc()) \
  .show(10)
```

# Windows Functions

Window functions let you compute rankings, running totals, and comparisons across groups without losing individual rows.

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number, rank, dense_rank, avg, col, lag

# Define a window specification partitioned by country and ordered by salary descending
window_spec = Window.partitionBy("country").orderBy(col("salary").desc())

# A. Ranking: Find top earners per country
df.withColumn("rank", rank().over(window_spec)) \
  .withColumn("dense_rank", dense_rank().over(window_spec)) \
  .withColumn("row_num", row_number().over(window_spec)) \
  .filter(col("rank") <= 2) \
  .select("country", "first_name", "last_name", "salary", "rank") \
  .show()

# B. Compare salary against country average without collapsing rows
window_country = Window.partitionBy("country")

df.withColumn("avg_country_salary", avg("salary").over(window_country)) \
  .withColumn("diff_from_avg", col("salary") - col("avg_country_salary")) \
  .select("first_name", "country", "salary", "avg_country_salary", "diff_from_avg") \
  .show(5)

# C. Lag function: Compare an employee's salary to the next higher salary in their country
df.withColumn("next_higher_salary", lag("salary", 1).over(window_spec)) \
  .select("country", "first_name", "salary", "next_higher_salary") \
  .show(5)
```

# Joining DataFrames

To practice joins, create a small lookup DataFrame for country tax rates or regional data:

```python
# Create a secondary DataFrame for country metadata
country_data = [
    ("Germany", "EUR", 0.35),
    ("Austria", "EUR", 0.32),
    ("Switzerland", "CHF", 0.22),
    ("France", "EUR", 0.30)
]
country_df = spark.createDataFrame(country_data, ["country", "currency", "tax_rate"])

# A. Inner Join
df.join(country_df, on="country", how="inner") \
  .select("first_name", "country", "salary", "currency", "tax_rate") \
  .show(5)

# B. Left Outer Join (keeps all records from people dataset even if country is missing)
df.join(country_df, on="country", how="left") \
  .select("first_name", "country", "tax_rate") \
  .show(5)

# C. Broadcast Join (optimizes performance by sending small table to all nodes)
from pyspark.sql.functions import broadcast

df.join(broadcast(country_df), on="country", how="inner").show(5)
```

# Advanced Column Transformations & UDFs

Work with complex conditionals, string manipulations, and Python functions wrapped as PySpark UDFs.

```python
from pyspark.sql.functions import udf, when, concat, lit, upper
from pyspark.sql.types import StringType

# A. Native Column Logic (Preferred for performance)
df_transformed = df.withColumn(
    "seniority",
    when(col("age") >= 50, "Senior")
    .when(col("age") >= 35, "Mid-Level")
    .otherwise("Junior")
).withColumn("full_name", concat(col("first_name"), lit(" "), upper(col("last_name"))))

df_transformed.select("full_name", "age", "seniority").show(5)

# B. Python User Defined Function (UDF) - used for custom logic when native functions aren't enough
def categorize_salary(salary):
    if salary > 100000:
        return "Executive Band"
    elif salary > 70000:
        return "Professional Band"
    else:
        return "Associate Band"

# Register UDF with return type
salary_udf = udf(categorize_salary, StringType())

df.withColumn("salary_band", salary_udf(col("salary"))) \
  .select("first_name", "salary", "salary_band") \
  .show(5)
```

# PySpark SQL Expressions

Register your DataFrame as a Temporary View to write raw ANSI SQL queries:

```python
# Create temporary SQL view from DataFrame
df.createOrReplaceTempView("people_view")

# Query using raw SQL syntax
query_res = spark.sql("""
    SELECT 
        country,
        job_title,
        COUNT(*) as total_count,
        ROUND(AVG(salary), 2) as avg_salary
    FROM people_view
    WHERE age > 25
    GROUP BY country, job_title
    HAVING COUNT(*) >= 1
    ORDER BY avg_salary DESC
""")

query_res.show(10)
```
